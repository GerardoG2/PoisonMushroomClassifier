```python
def load_and_preview_data(file_path='../Data/raw_data.csv', header=True):
    """
    Load and process mushroom data from CSV file.
    
    Parameters:
    file_path (str): Path to the CSV file
    header (bool): Whether the file has a header row
    
    Returns:
    pandas.DataFrame: Processed mushroom data
    """
    # Read CSV with semicolon separator (not comma)
    df = pd.read_csv(file_path, sep=';')
    
    # Display basic info about the dataset
    print(f"Dataset shape: {df.shape}")
    print(f"Columns: {list(df.columns)}")
    print(f"Missing values per column:")
    print(df.isnull().sum())
    
    return df
```

```python
def display_class_distribution(df):
    """
    Display the class distribution of the mushroom dataset.
    
    Parameters:
    df (pandas.DataFrame): The mushroom dataset
    
    Returns:
    None: Prints the distribution and percentages
    """
    print("Class Distribution:")
    print("=" * 30)
    
    # Count values in the 'class' column
    class_counts = df['class'].value_counts()
    print(f"Total samples: {len(df)}")
    print()
    
    for class_label, count in class_counts.items():
        percentage = (count / len(df)) * 100
        class_name = "Poisonous" if class_label == 'p' else "Edible"
        print(f"{class_name} ({class_label}): {count:,} samples ({percentage:.2f}%)")
    
    print()

    print("Class Balance:")
    ratio = class_counts.max() / class_counts.min()
    print(f"Imbalance ratio: {ratio:.2f}:1")
    
    return class_counts
```

## original preprocessing 

**Impute train and test data**
* We will use KNN imputation on the columns with less than 50% missing data.
* We use k = 5.

```python
# standardize numeric columns
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
numeric_cols = ['cap-diameter', 'stem-height', 'stem-width']

# standardize train data
std_train_df = train_df.copy()
std_train_df[numeric_cols] = scaler.fit_transform(train_df[numeric_cols])
std_train_df.describe()

# standardize test data
std_test_df = test_df.copy()
std_test_df[numeric_cols] = scaler.transform(test_df[numeric_cols])
std_test_df.describe()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline

def knn_impute_categorical(df, target):
    """
    Impute missing categorical values using KNN classifier.

    Target: column to impute
    df: dataframe with missing values
    """

    mask = df[target].notna()  # rows where column does not have missing values

    # build train_columns based on columns without nulls 
    _X_train_tmp = df.loc[mask].drop(columns=[target])
    train_columns = _X_train_tmp.columns[_X_train_tmp.isnull().sum() == 0].tolist() # train on columns without nulls

    # training data where column is not null
    X_train = df.loc[mask, train_columns]
    y_train = df.loc[mask, target]

    # predict(impute) on rows where column ARE NULL
    X_missing = df.loc[~mask, train_columns]

    # define numeric and categorical columns
    num_cols = ['cap-diameter','stem-height','stem-width']
    cat_cols = [c for c in train_columns if c not in num_cols + [target]]

    # create preprocessor
    pre = ColumnTransformer([
        ("num", StandardScaler(), num_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), cat_cols),
    ])

    # knn classifier pipeline
    clf = Pipeline([
        ("pre", pre),
        ("knn", KNeighborsClassifier(n_neighbors=5, weights="distance"))
    ])

    # fit knn classifier and predict (IMPUTE) missing values
    clf.fit(X_train, y_train)
    df.loc[~mask, target] = clf.predict(X_missing)  # fill in missing cap-surface values

    return df
```

```python


# categorical columns with missing values
categorical_cols_with_nans = ['cap-surface', 'gill-attachment', 'gill-spacing', 'ring-type']

# impute missing values iteratively using knn for categorical columns with missing values
imputed_train_df = std_train_df.copy()
imputed_test_df = std_test_df.copy()

for col in categorical_cols_with_nans:
    imputed_train_df = pd.DataFrame(knn_impute_categorical(imputed_train_df, col))
    imputed_test_df = pd.DataFrame(knn_impute_categorical(imputed_test_df, col))

print("missing values in train data:")
print(imputed_train_df.isnull().sum())
print()
print("missing values in test data:")
print(imputed_test_df.isnull().sum())
```

```python
# encode features
from sklearn.preprocessing import OneHotEncoder

# do not encode class
imputed_train_df_no_class = imputed_train_df.drop(columns=['class'])
imputed_test_df_no_class = imputed_test_df.drop(columns=['class'])

# get categorical columns
cat_columns = imputed_train_df_no_class.select_dtypes(include=['object']).columns.tolist()

encoder = OneHotEncoder(categories = 'auto', sparse_output=False, handle_unknown='ignore')

# encode train data
encoded_train = imputed_train_df_no_class[cat_columns]
encoded_train = encoder.fit_transform(encoded_train)
# encode test data
encoded_test = imputed_test_df_no_class[cat_columns]
encoded_test = encoder.transform(encoded_test)

# get feature names 
feature_names = encoder.get_feature_names_out(cat_columns)

# encoded dataframes
encoded_train__featuresdf = pd.DataFrame(
    encoded_train,
    columns=feature_names
)

encoded_test__features_df = pd.DataFrame(
    encoded_test,
    columns=feature_names
)
encoded_test__features_df
```

```python
# encode class column
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

# encode class in train set
train_class = imputed_train_df['class']
train_class_encoded = le.fit_transform(train_class)
# convert to series
train_class_encoded = pd.Series(train_class_encoded, name='class')

# encode class in test set
test_class = imputed_test_df['class']
test_class_encoded = le.transform(test_class)
# convert to series
test_class_encoded = pd.Series(test_class_encoded, name='class')
train_class_encoded

```

```python
# concatenate encoded features with class column (train set)
final_train_df = pd.concat(
    [encoded_train__featuresdf, pd.DataFrame(train_class_encoded, columns=['class'])],
    axis=1
)


# concatenate encoded features with class column (test)
final_test_df = pd.concat(
    [encoded_test__features_df, pd.DataFrame(test_class_encoded, columns=['class'])],
    axis=1
)

final_train_df
```
